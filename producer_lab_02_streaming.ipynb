```python
import requests
import json
```

```python
def sourceClient():
        response = requests.get("http://api.open-notify.org/iss-now.json")
        registro = json.loads(response.content)
        return registro
```

```python
sourceClient()
```

```python
def binary(registro):
        registroBin = json.dumps(registro)
        return registroBin
```

```python
type(binary(sourceClient()))
```

```python
from kafka import KafkaProducer

def writer(registroBin):
        producer = KafkaProducer(bootstrap_servers="kafka:9092")
        producer.send("proyect-final-topic", registroBin.encode())
        producer.flush()        
```

```python
import time

for i in range(1, 10):
        print("Iteracion: "+str(i))
        registro = sourceClient()
        registroBin = binary(registro)
        writer(registroBin)
        time.sleep(3)
```
